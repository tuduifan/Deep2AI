```python
# 背下步骤
# 1.生成数据集
# 2.读取数据集
# 3.定义模型
# 4.初始化模型参数
# 5.定义损失函数
# 6.定义优化算法
# 7.训练（①计算损失，②反向传播，③更新参数）
```

```python
# 1.生成数据集
import torch
from torch.utils import data
import numpy as np

true_w = torch.tensor([2,-3.4])
true_b = 4.2
features = torch.normal(0,1,(1000,len(true_w)))
labels = (torch.matmul(features,true_w)+true_b).reshape(-1,1)
features,labels
```

```python
# 2.读取数据集
batch_size = 10
dataset = data.TensorDataset(features,labels)
data_iter = data.DataLoader(dataset,batch_size,shuffle=True)
dataset,dataset[0],data_iter,iter(data_iter)
```

```python
# 3.定义模型
from torch import nn

net = torch.nn.Sequential(nn.Linear(2,1))
```

```python
# 4.初始化模型参数
net[0].weight.data.normal_(0,0.01)
net[0].bias.data.fill_(0)
net[0].weight.data,net[0].bias.data,net[0].weight.shape
```

```python
# 5.定义损失函数
loss = nn.MSELoss()
help(nn.MSELoss)
```

```python
# 6.定义优化算法
trainer = torch.optim.SGD(net.parameters(),lr=0.03)
net.parameters
```

```python
# 7.训练
num_epochs = 3
for epoch in range(num_epochs):  # 在epoch循环下需遍历一次整个数据集
    for X,y in data_iter:  # 采用SGD方法
        l = loss(net(X),y)
        trainer.zero_grad()
        l.backward()
        trainer.step()
    l = loss(net(features),labels)
    print(f"epoch:{epoch}, loss:{l:.10f}")
```
